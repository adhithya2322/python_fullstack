# Functions

```python
# Def: A function is block of  reusable code that perform a specific task.
def func():
    print("Welcome to functions")
    print("Hello World")   
func() ## Calling a function    
func()
```

```python
# Example 2: passing one parameter in the function
def func(val1, val2):
    print("Welcome to functions", val1)
    print("Hello World", val2)   
func(100,200) ## Calling a function    
#func()
```

```python
# Exampl-3: nested function : function inside function
def outer():
    print(" I am outer function")
    def inner():
        print(" iam inner function")
    inner()
outer()
```

```python
def Building():
    print("We are here")
    def room():
        print("pavan iam here")
    room() # calling from inside the Building function
Building()
```

# Default parameters

```python
# assign a default value to the "var" variable
def fun(var = "adhithya"):
    print(var)
    
fun(9)
```

```python
def func2(x=10, y=20):
    print(x+y)
func2()
func2(100)
func2(100,200)
```

# passing arbitrary number of argumenrs using *args

```python
def func(*args):
    for i in args:
        print(i)
func()
func("Rohith")
func("Rohith", "Mahesh")
```

# passing arbitrary key-value based arguments using *args

```python
def pavan(**GF):
    for key in GF:
        print(key, ":", GF[key])
pavan(chikki=150, Minki=250) 
pavan(mahesh=150)
pavan()
```
