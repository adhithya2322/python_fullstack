# Lambda Function

```python
# Def: A Lambda Function is a single line of code that us used to perform
        a short task without defining a full function with def. 
# A Lambda function do not have any name
# Although Lambda function can recieve arguments and process them 
# lambda function are popularly used inside the other function 
```

```python
obj = lambda a : a + 100
print(obj(10))
```

```python
obj = lambda a,b: a + b
print(obj(100,200))
```

```python
obj1 = lambda a,b,c : a*b*c
val = obj1(1,2,3)
print(val)
```

```python
# TO perform some small task
def fun(p):
    return lambda a : a**p
obj1 = fun(10)
print(obj1(2))
print(obj1(3))
print(obj1(4))
```

```python
# Using Lambda inside the other function 
def fun(p):
    obj = lambda a : a**p if a%2==0 else a+p
    return obj
obj = fun(2)
print(obj(3))
print(obj(4))
```

```python
# sort the below given list of tuple 
students = [("pavan", 82), ("mahesh", 88), ("ajith", 80)]
#obj = lambda sorted (students)
sorted_data=sorted(students, key = lambda a : a[1])
print(sorted_data)
```

```python
students = {"pavan":82, "mahesh": 88, "ajith": 80}
data = students.items()
print(data)

sorted_data=sorted(data, key = lambda a : a[1])
print(sorted_data)
```

# Recursive Function 

```python
# def: The Function calling itself directly or indirectly . 
# Termination condition is must recursion 
# recurisve call, we must change to increment or decrement 
```

```python
def foo(i):
    if i <1:
        return
    print(i)
    foo(i-1)
foo(6)
```

```python
# Example 2 
def foo(i):
    if i <1:
        return
    foo(i-1)
    print(i)
 
foo(6)
```

```python
# Example 2 
def foo(i):
    if i <1: # termination condition or base condition 
        return
    print(i)
    foo(i-1) # recursive call
    print(i)
 
foo(6)
```
