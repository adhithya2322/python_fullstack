# 1. Arthimetic Operators ( +, -, *, /, //, %, **)


```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** b)
print(a % b)
print(a // b)
```

# 2. Bitwise operators ( <<, >>, ~, &, |, ^)

```python
#Bitwise operators works on binary number
#Bitwise operators are faster than arithmetic operators
#Bitwise operators performs many operations in easy way due to binary number systems.
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

```python
a = 10
b = 3
c = 2

print(a & b & c)
```
