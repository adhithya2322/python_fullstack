```python
# Basic interview Questions
# Q1. Set is mutable or immutable in Python? Explain.
#  is a collection of unique elements. It does not allow duplicate values and is unordered. Sets are mutable, meaning you can add or remove elements after creation. They are commonly used for membership testing, removing duplicates from a list, and performing mathematical set operations like union, intersection, and difference.

# Q2. What is the time complexity of adding an element to a set in Python?
# The time complexity of adding an element to a set in Python is O(1) on average. it is because sets are implemented using hash tables, which allow for constant-time average complexity for insertions. However, in the worst case, the time complexity can be O(n) if there are many hash collisions.

# Q3. Dulplicate elements are allowed in a set? Explain.
# No, duplicate elements are not allowed in a set. When you try to add a duplicate element to a set, it will be ignored, and the set will remain unchanged. This property of sets ensures that each element is unique within the collection. If you need to store multiple occurrences of an element, you should use a list or another data structure that allows duplicates.

# Q4. Heterogeneous elements are allowed in a set? Explain.
# Yes, heterogeneous elements are allowed in a set. A set can contain elements of different data types, such as integers, strings, floats, and even tuples. However, all elements in a set must be hashable (i.e., they must have a hash value that does not change during their lifetime). Mutable types like lists and dictionaries cannot be added to a set because they are not hashable.

# Q5. Set is ordered or unordered in Python? Explain.
# Sets in Python are unordered collections of unique elements. This means that the elements in a set do not have a specific order, and their arrangement can change over time. When you iterate over a set, the order of elements may appear random and can vary between different runs of the program. This unordered nature is due to the underlying implementation of sets using hash tables, which prioritize efficient membership testing and uniqueness over maintaining order. If you need an ordered collection, you can use a list or an OrderedDict (from the collections module) instead.
```

```python
# Delete Operation on set (remove(), Discard(), pop())
```

```python
dinner_set={"dosa","idli","vada","poori"}

# remove() funncation delete the one item whose name is provided
# remove() will raaise on ERROR if ithem don't exist in the set.

dinner_set.remove("idli")
dinner_set.discard("vada")
print(dinner_set)
```

```python
# remove() will raaise on ERROR if ithem don't exist in the set.
dinner_set.remove("vada")
print(dinner_set)
```

```python
# discard DONT raise an error if the item doesn't exist in the set, it will simply do nothing.
dinner_set.discard("vada")
print(dinner_set)
```

```python
    # pop() will delete any random item from the set and return the deleted item. If the set is empty, it will raise a KeyError.
dinner_set.pop()
print(dinner_set)
```
