## updating List Items

```python
list_politician = ["Vijay", "CBN", "Pavan", "Jailalitha", "Balayya"]
list_politician[1] = "Dance with"
print(list_politician)
```

```python
list_politician = ["Vijay", "CBN", "Pavan", "Jailalitha", "Balayya"]
list_politician[3:5] = ["Dance with"]
print(list_politician)
```

```python
list_politician[1:3]= ["Rahul", "Kejriwal", "Yogi", "Mamta"]
print(list_politician)
```

```python
list_politician = ["Rahul", "Kejriwal", "Yogi", "Mamta"]
list_politician[2:-1]= ["Dance with"]
print(list_politician)
```

```python
list_politician = ["Rahul", "Kejriwal", "Yogi", "Mamta"]
list_politician[-5:-1]= ["Dance with"]
print(list_politician)
```

### Delete item from list

## Delete item from list(remove(), pop(), clear(), del()

```python
food = ['pizza', 'burger', 'pasta', 'salad', 'tacos']
print(food)
```

```python
food.remove("burger")
print(food)
```

```python
food.pop(1)
print(food)
```

```python
food.append("biryani")
print(food)
```

```python
del food[1:4]
print(food)
```

```python
food.clear()
print(food)
```

```python
del food
print(food)
```
