# Enumerate Function

--> enumerate() function add the counter (index) with list / tuple items.

```python
fruits_names=["mamidipandu", "PanasaPandu", "JamaPandu"]
 
for index,name in enumerate(fruits_names, start=1):  # By defualt it is start from ZERO but we can give the starting point
    print(index, name)
```

```python
# Enumerate with Tuple
tu=("Mahesh","adi",101,102, 5.0,9999)
for index, name in enumerate(tu, start=101):
    print(index, name)

```

# Zip() Function

--> zip() function is used to combine/zip mutiple list/tuple items, element-wise
--> The Zip() need equal length of list/tuple

```python
hero=["Adi", "malikarjun", "pavan kalyan", "balaji"]
heroines=["Anushka", "samantha",  "rashmika", "Kajal"]
movie=["x","y","z","h"]
price=[20,30,40,50]

pairs=zip(hero,heroines, movie, price)
print(list(pairs))
```

```python
# --> The Zip() need equal length of list/tuple, if not it will not print the last one , it will discard it
hero=["Adi", "malikarjun", "pavan kalyan", "balaji"]
heroines=["Anushka", "samantha",  "rashmika", "Kajal"]
movie=["x","y","z","h"]
price=[20,30,40]

pairs=zip(hero,heroines, movie, price)
print(list(pairs))
```

# Decorators

Def:--> A decorator is a function that take another function as input, adds extra functionality and return a modified function, without changing the original function code.



```python
def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
        print("add cherry")
        print("add chocolate")
        
    return wrapper_fun
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
decorate_fun=dec_fun(cake)
decorate_fun()
```

```python
def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
        print("add cherry") #  Extra functionality
        print("add chocolate")
        
    return wrapper_fun
@dec_fun
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```

```python
# Multiple decorator for same function

def dec_fun(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients to make cake")
        othfun()
    return wrapper_fun
def dec_fun2(othfun): # fun --> is cake()
    def wrapper_fun(): # wrapper function add the extra functionality
        othfun()
        print("add cherry") #  Extra functionality
        print("add chocolate")
        
    return wrapper_fun

@dec_fun
@dec_fun2
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```

```python
# Example 1: Structure of class
# class is the keyword in python, food is the name of class
# class variable : The value of class variable remamins Same for all the objects 
# class variable value cannot be change by object 
# class variable value can be change by class name 
class food:
    food_name = "Biryani" # class Variable

obj3 = food()
obj3.food_name = "Pappu" # Here  obj3, is creating its own local variable with same name
print(obj3.food_name) # printing local variable to obj3

obj=food() #Creating the object of food() class
print(obj.food_name) # accessing the class variable , Biryani
food.food_name="Mutton"

obj2=food()
print(obj2.food_name) # class object obj2 can access the class data
# acess the class variable with the name 
print(food.food_name)

```

```python
# Example-2: Constructor function 
# constructor function automatically get called when object is created.
# constructor is created by __init__
# Constructor function is used for initializing the variable 
# Instance variable value id DIFFERENT for every object  
# class variable value is same for all the objects 
# self is a variable  which points to the current object 
class Bike:
    No_types = 2 # class Variable
    def __init__(self, name, colour, price):
        self.bike_name = name
        self.bike_colour = colour
        self.bike_price = price
        print(self.bike_name, end ="-")
        print(self.bike_colour, end=" ")
        print(self.bike_price)

obj = Bike("RE", "Red", "250000")
obj1 = Bike("KTM", "Black", 300000)
```
