### Encapsulation 

```python
# Def: Encapsulation refers to the binding of data (attributes) 
#  and functions into one unit(class).

# Encapsulation restrict the direct access of the data. 
# We can access the variable/data/attribute through object outside the class
# We can access the variable/data/attribute through function defined inside the class
```

```python

class Bike:
    No_tyres = 2 # class Variable / attribute / data
    def __init__(self, name, colour, price):
        self.bike_name = name  # Instance Variable
        self.bike_colour = colour
        self.bike_price = price
        print(self.bike_name, end ="-")
        print(self.bike_colour, end="-")
        print(self.bike_price)

  
    def display(self):
        print(self.No_tyres,self.bike_name, self.bike_colour, self.bike_price)
obj = Bike("RE", "Red", 250000)
obj1 = Bike("KTM", "Black", 300000)
obj1.display()
```

# Access specifers / type of attributes in encapsulation

```python
# 1. public: these variables can be accessed outside the class 
        #public variable do not start with underscore(_)
# 2. protected: these variable can be accessed only inside  and child class
        # protected varaible NoT INTENDED to access outside the class 
        # protecte vaarable start with SINGLE Underscore
#3. private These variable can be accessed only inside the class 
        # NoT INTENDED to access outside/child class 
```

```python
class marvels:
    def __init__(self, name, power, weapon):
        self.Name = name # public instance variable
        self._Power = power # protected instance variable 
        self.__Ayudham = weapon # private instance variable
    def show (self):
        print(self.Name, self._Power, self.__Ayudham)
obj = marvels("Thar", 1000, "Hammer")
obj.show()
print(obj.Name)
print(obj._Power) # Protected variable can be accessed outside class but you should not access
 # In python public, protected, privste concept is used just for sign/Warning
print(obj._marvels__Ayudham) # It is private variable canot access outside the class
```
