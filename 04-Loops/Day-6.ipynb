```python
i = 1
while (i <= 10):
    i += 1
    print(i)
    
```

```python
# Example 2: break statment
# break statement helps to come out of the loop

i = 1 
while i <= 10:
    print(i)
    if i == 5:
        break
    i += 1
```

```python
i = 1 
while i <= 10:
    i += 1
    print(i)
    if i == 5:
        break
    
```

```python
# Example 2 : Continue statement
# continue will take control to the start of the loop 

i = 1 
while i < 10:
    i += 1
    if i < 5:
        continue
    print(i)
```

```python
# else statement with while loop
# else statement executes only if loop is successfully completed

i = 1
while i<=10:
    print(i)
    i+=1
    if i > 5:
        break
else:
    print("loop is sucessfully executed")
```

```python
# Example - 5 : Pass statement
# pass statement is a placeholder for future code 

i =0
while i<=10:
    if i==5:
        pass
    else:
        print(i)
    i+=1
```
