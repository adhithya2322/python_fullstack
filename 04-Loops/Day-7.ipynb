```python
# Example 1: break statement
# break statement will bring the control out of the loop

for num in range(1,6):
    if num == 3:
        break
    print(num)
```

```python

for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# Example 6 nested loop
#for every value of outerloop , inner loop will completely execute
for i in range(1, 4):
    for j in range(1, 3):
        print(i,j)
        
```

```python
for i in range(3, 6):
    for j in range(2, 5):
        print(i,j)
        
```

```python
for i in range(3, 6):# n = 3 times
    for j in range(2, 7): # n = 5 times  # time complexity = O(n * m)
        print(i,j)
        
```

# 4. Indexing & Slicing

```python
#string [start:end:step]
#start = 0
#end = n-1
#stepsize = 1
Heroine = "Samantha"

print(Heroine[4])
print(Heroine[-4])
print(Heroine[-6])
print(Heroine[0:5])
print(Heroine[0:3] + Heroine[-7:-2])
print(Heroine[0:7:3])
print(Heroine[0:8])
print(Heroine[::2])
print(Heroine[::])
print(Heroine[7:1:-1])
```

```python
print(Heroine[-1])
print(Heroine[-6])
print(Heroine[-1:-6:-1])
print(Heroine[-7::])
```

```python
print(Heroine[0:3] + Heroine[-5:-1])
```

```python
print(Heroine[::-1])
```

```python
#reverse the string
s="racecar"
print(s[::-1])
```
