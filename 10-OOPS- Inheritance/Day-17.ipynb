# Multilevel Inheritance 

```python
# son/daughter : father :> grandfather :
```

```python
class granter:
    def Gold(self):
        print("1000 KG")

class grandfather(granter):
    def Land(self):
        print("500 Acr")
        
class father(grandfather):
    def car(self):
        print("Lambo")
class son(father):
    def Bike(self):
        print("RE Bike")
obj_son = son()
obj_son.Bike()
obj_son.car()
obj_son.Land()
obj_son.Gold()
```

# Hybrid Inheritance

```python
class A:
    A_var = "A Class"
    
class B(A):
    B_var = "B Class"
class C(B):
    C_var = "C Class"
class D(B):
    D_var = "D Class"
class E(C,D):
    E_var = "E Class"
obj_E = E()
print(obj_E.A_var)
```

```python

# Example 2 : By constructor
class A:
    def __init__(self):
         print("Constructor at A")
    def classA(self):
        print("Class A")
      
class B(A):
    def __init__(self):
     print("Constructor at B")
    def classB(self):
        print("Class B and Parent: A")
class C(B):
    def classC(self):
        print("Class C and Parent: B")
        

class D(B):
    def classD(self):
        print("Class C and Parent: B")
class E(C,D):
    def classE(self):
        print("Class C and Parent: C,D")

objE=E()
objE.classE()
objE.classA()
```
