# 1. Single Inheritance: One parent one child

```python
# Example-1  Structure of single inheritance
# parent
class hyd_food:
    taste = "very very spicy"

    def Biryani(self):
        print("Very Very tasty Biryani")
#child class 
class PistaHouse(hyd_food):
    pass
obj_child = PistaHouse()  # It is the object of child class

# child class object can access the properties & function of parent class.
print(obj_child.taste)
obj_child.Biryani()
```

```python
# Eample-2 super() function
#super() : used to access the parent properties and function from 
#            inside of the child class 


class hyd_food:
    taste = "very very spicy"

    def Biryani(self):
        print("Very Very tasty Biryani")
#child class 
class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)
obj_child = PistaHouse()  # It is the object of child class

# child class object can access the properties & function of parent class.
print(obj_child.taste)
obj_child.display()
```

```python
class hyd_food:
    taste = "very very spicy"

    def __init__(self):
        print("Yes it is very tasty")

#child class 
class PistaHouse(hyd_food):
    def __init__(self):
        print(" i like biryani")
obj_child = PistaHouse()  
obj_child=hyd_food()
obj_child.taste
# if the class constructor function is not defined 
# then child class object will call automatically parent class function 
```

# 2 Multiple Inheritance: Many parents one child

```python
# Example-1: Structure for Mutiple inheriteance
class A:
    def display(self):
        print("I am a A class")
class B:
    def show(self):
        print("I am B class")
class C(A,B):
    pass
obj1=C()
obj1.display()
obj1.show()
```

```python
# Expample-2:- sequecne in which parent are assgin to other
class A:
    def display(self):
        print("I am a A class")
class B:
    def display(self):     # same function name
        print("I am B class")
class C(A,B):  # but here is the order for priting
    pass
obj1=C()
obj1.display()
```
